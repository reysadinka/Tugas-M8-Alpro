```python
pelamar = [("Budi", 3.2, 5), ("Siti", 3.6, 3), ("Dodi", 3.8, 1)]
print("=== HASIL STATUS PELAMAR ===")
for nama, ipk, semester in pelamar:
    if ipk >= 3.5 and semester > 2:
        status = "Diterima"
    else:
        status = "Ditolak"
    print(f"{nama}: {status}")
```

```python
karyawan = [("Joko", "IT", 7000), ("Rina", "HR", 5000), ("Toni", "Data", 8000), ("Lia", "IT", 6500)]
total_gaji = 0
for nama, dept, gaji in karyawan:
    if dept == "IT" or dept == "Data":
        total_gaji += gaji
print(total_gaji)
```

```python
penjualan = [("Kopi", 60, False), ("Teh", 30, False), ("Boba", 80, True)]
print("=== DAFTAR MINUMAN LAKU KERAS MURNI ===")
for nama, terjual, promo in penjualan:
    if terjual > 50 and not promo:
        print(f"Minuman: {nama} (Terjual: {terjual} cup)")
```
